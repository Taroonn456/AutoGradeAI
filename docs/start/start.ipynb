```
# ============================================================
# CELL 5: MOLECULAR ENCODING
# Morgan Fingerprints (radius=2, 2048 bits)
#   → StandardScaler
#   → PCA to 4 features (one per qubit)
#   → Scale to [-π, π] for AngleEmbedding
# ============================================================

N_QUBITS = 4      # 4 qubits = 4 input features per molecule
N_PCA_COMPONENTS = N_QUBITS

def smiles_to_morgan_fp(smiles, radius=2, nbits=2048):
    """
    Convert SMILES string → Morgan circular fingerprint (bit vector).
    Morgan FP captures local molecular environment around each atom.
    radius=2 is equivalent to ECFP4 used in drug discovery literature.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return classical_np.zeros(nbits)
    fp = GetMorganFingerprintAsBitVect(mol, radius=radius, nBits=nbits)
    return classical_np.array(fp)

print("🔬 Generating Morgan Fingerprints (ECFP4, 2048 bits)...")
fp_matrix = classical_np.array([
    smiles_to_morgan_fp(smi) for smi in df['Smiles']
])
labels = df['label'].values

print(f"✅ Fingerprint matrix shape: {fp_matrix.shape}")

# --- Train/Test Split BEFORE encoding (prevent data leakage) ---
X_train_fp, X_test_fp, y_train, y_test = train_test_split(
    fp_matrix, labels,
    test_size=0.20,
    random_state=SEED,
    stratify=labels
)
print(f"\nTrain size: {len(X_train_fp)} | Test size: {len(X_test_fp)}")

# --- StandardScaler (fit on train only) ---
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_fp)
X_test_scaled  = scaler.transform(X_test_fp)

# --- PCA: 2048 → 4 features (fit on train only) ---
pca = PCA(n_components=N_PCA_COMPONENTS, random_state=SEED)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca  = pca.transform(X_test_scaled)

print(f"\n📉 PCA variance explained: {pca.explained_variance_ratio_.sum():.3f} ({N_PCA_COMPONENTS} components)")
print(f"Per-component variance: {pca.explained_variance_ratio_}")

# --- Scale features to [-π, π] for AngleEmbedding ---
def scale_to_angle(X):
    """Normalize each feature to range [-pi, pi] for qubit rotation angles."""
    X_min = X.min(axis=0)
    X_max = X.max(axis=0)
    X_norm = (X - X_min) / (X_max - X_min + 1e-8)  # [0, 1]
    return X_norm * 2 * classical_np.pi - classical_np.pi  # [-pi, pi]

X_train_q = scale_to_angle(X_train_pca)
X_test_q  = scale_to_angle(X_test_pca)

# Convert to PennyLane-compatible numpy
X_train_q = classical_np.array(X_train_q)
X_test_q  = classical_np.array(X_test_q)
y_train   = classical_np.array(y_train)
y_test    = classical_np.array(y_test)

print(f"\n✅ Quantum feature matrix (train): {X_train_q.shape}")
print(f"Feature range: [{X_train_q.min():.3f}, {X_train_q.max():.3f}] (should be in [-π, π])")
```

```
# ============================================================
# CELL 6: QUANTUM DEVICE SETUP
# Using Google Cirq simulator via PennyLane-Cirq plugin
# cirq.simulator = Cirq's built-in statevector simulator
# ============================================================

# Number of qubits = number of features after PCA
N_QUBITS = 4
N_LAYERS = 3   # VQC depth: 3 strongly entangling layers

# ---- Define Cirq device ----
# cirq.simulator uses Cirq's native statevector simulation
# Fully compatible with PennyLane's gradient pipeline (parameter-shift rule)
try:
    dev = qml.device('cirq.simulator', wires=N_QUBITS)
    print(f"✅ Quantum device: cirq.simulator")
    print(f"   Wires (qubits): {N_QUBITS}")
    print(f"   VQC Layers: {N_LAYERS}")
except Exception as e:
    print(f"⚠️  Cirq backend not available, falling back to default.qubit")
    print(f"   Error: {e}")
    dev = qml.device('default.qubit', wires=N_QUBITS)
    print("✅ Using PennyLane default.qubit simulator")

# --- Circuit architecture parameter count ---
# StronglyEntanglingLayers: 3 params per qubit per layer
n_params = N_LAYERS * N_QUBITS * 3
print(f"\n📐 Trainable parameters: {n_params}")
print(f"   Formula: {N_LAYERS} layers × {N_QUBITS} qubits × 3 rotation angles")
```

```
# ============================================================
# CELL 7: VARIATIONAL QUANTUM CIRCUIT (VQC) DEFINITION
#
# Architecture:
#   1. AngleEmbedding  → encode 4 molecular features as Ry rotations
#   2. StronglyEntanglingLayers (3) → parameterized Ry, Rz + CNOT entanglement
#   3. Measurement → expectation value of PauliZ on qubit 0
#   4. Sigmoid → binary classification probability
#
# Loss: Binary Cross-Entropy
# Optimizer: Adam (quantum parameter-shift gradient)
# ============================================================

@qml.qnode(dev, interface='autograd', diff_method='parameter-shift')
def vqc_circuit(inputs, weights):
    """
    VQC circuit for binary molecular activity classification.

    Args:
        inputs  : shape (4,) — scaled molecular features from PCA
        weights : shape (N_LAYERS, N_QUBITS, 3) — trainable rotation angles
    Returns:
        float: expectation value <Z₀> ∈ [-1, +1]
    """
    # ---- ENCODING LAYER ----
    # AngleEmbedding: maps each feature xᵢ → Ry(xᵢ) rotation on qubit i
    # This encodes classical molecular information into quantum state
    qml.AngleEmbedding(inputs, wires=range(N_QUBITS), rotation='Y')

    # ---- VARIATIONAL LAYERS ----
    # StronglyEntanglingLayers: each layer applies:
    #   Ry(θ) + Rz(φ) on each qubit, followed by
    #   CNOT entanglement gates between neighboring qubits
    qml.StronglyEntanglingLayers(weights, wires=range(N_QUBITS))

    # ---- MEASUREMENT ----
    # Measure expectation value of Pauli-Z on qubit 0
    # Result ∈ [-1, +1]; we map this to [0, 1] for binary classification
    return qml.expval(qml.PauliZ(0))


def sigmoid(x):
    """Map VQC output [-1, +1] → probability [0, 1]"""
    # Use PennyLane's numpy for differentiability
    return 1 / (1 + np.exp(-x * 5))  # scale factor 5 for sharper decision


def predict_proba(X, weights):
    """Get probability predictions for a batch of molecules."""
    probs = []
    for x in X:
        raw = vqc_circuit(x, weights)
        # No float() needed here, sigmoid handles ArrayBox
        prob = sigmoid(raw)
        probs.append(prob)
    return classical_np.array(probs)


def predict(X, weights, threshold=0.5):
    """Binary predictions (0 or 1)."""
    probs = predict_proba(X, weights)
    return (probs >= threshold).astype(int)


# ---- Draw circuit (first molecule as example) ----
print("🔬 VQC Circuit Architecture:")
weights_example = classical_np.zeros((N_LAYERS, N_QUBITS, 3))
print(qml.draw(vqc_circuit)(X_train_q[0], weights_example))
print(f"\n✅ VQC defined on cirq.simulator | {n_params} trainable parameters")
```

```
# ============================================================
# CELL 8: COST FUNCTION — Binary Cross-Entropy
#
# L = -1/N * Σ [y·log(p) + (1-y)·log(1-p)]
#
# This is the standard BCE loss for binary classification.
# We compute it over mini-batches for efficiency.
# Gradients computed via parameter-shift rule (quantum-native).
# ============================================================

def binary_cross_entropy(weights, X_batch, y_batch):
    """
    Binary Cross-Entropy loss for VQC classification.

    Args:
        weights : VQC trainable parameters (N_LAYERS, N_QUBITS, 3)
        X_batch : batch of molecular feature vectors, shape (batch, 4)
        y_batch : true binary labels, shape (batch,)
    Returns:
        float: average BCE loss over batch
    """
    eps = 1e-8  # prevent log(0)
    total_loss = np.array(0.0) # Initialize as PennyLane numpy array for differentiability

    for x, y in zip(X_batch, y_batch):
        raw = vqc_circuit(x, weights)
        p = sigmoid(raw) # Removed float()
        p = np.clip(p, eps, 1 - eps) # Changed to np.clip
        loss = -(y * np.log(p) + (1 - y) * np.log(1 - p)) # Changed to np.log
        total_loss += loss

    return total_loss / len(X_batch)


def cost_fn(weights, X_batch, y_batch):
    """Wrapper for optimizer (must accept weights as first arg)."""
    return binary_cross_entropy(weights, X_batch, y_batch)


print("✅ Cost function defined: Binary Cross-Entropy")
print("   Gradient method: Parameter-Shift Rule (quantum-native)")
print("   Formula: L = -1/N * Σ [y·log(p) + (1-y)·log(1-p)]")

# Quick sanity check
weights_init = classical_np.random.uniform(
    low=-classical_np.pi,
    high=classical_np.pi,
    size=(N_LAYERS, N_QUBITS, 3)
)
# Convert to PennyLane-compatible numpy array if gradients are needed later
weights_init = np.array(weights_init, requires_grad=True)

test_loss = cost_fn(weights_init, X_train_q[:3], y_train[:3])
print(f"\n🧪 Sanity check loss (random weights, 3 samples): {test_loss:.4f}")
print(f"   (Expected ~0.693 = ln(2) for random binary classifier)")
```

```
# ============================================================
# CELL 9: TRAINING LOOP
#
# Optimizer: PennyLane Adam (quantum-aware)
# Epochs: 60
# Batch size: 16 (mini-batch SGD)
# Learning rate: 0.01
#
# NOTE: Due to quantum circuit simulation overhead,
# runtime is ~15-40 min on CPU. Reduce EPOCHS or BATCH_SIZE
# for faster testing. GPU does NOT accelerate circuit sim.
# ============================================================

EPOCHS     = 60
BATCH_SIZE = 16
LR         = 0.01
LOG_EVERY  = 5     # print progress every N epochs

# Subsample for faster prototyping (remove for full run)
# Use first 200 training samples for quick iteration
# Comment out these 2 lines for full dataset training
SUBSAMPLE = 200
X_tr = X_train_q[:SUBSAMPLE]
y_tr = y_train[:SUBSAMPLE]
# For full training: X_tr, y_tr = X_train_q, y_train

# ---- Initialize optimizer ----
optimizer = qml.AdamOptimizer(stepsize=LR)

# ---- Initialize weights (Xavier-style for quantum) ----
weights = classical_np.random.uniform(
    low=-classical_np.pi / 4,
    high=classical_np.pi / 4,
    size=(N_LAYERS, N_QUBITS, 3)
)
# Convert to PennyLane-compatible numpy array to enable gradient tracking
weights = np.array(weights, requires_grad=True)

# ---- Training history ----
history = {'loss': [], 'train_f1': [], 'epoch_time': []}

n_batches = len(X_tr) // BATCH_SIZE
print(f"🚀 Training VQC on cirq.simulator")
print(f"   Samples: {len(X_tr)} | Epochs: {EPOCHS} | Batch: {BATCH_SIZE}")
print(f"   Batches/epoch: {n_batches} | LR: {LR}")
print(f"   Trainable params: {weights.size}")
print("-" * 55)

total_start = time.time()

for epoch in range(1, EPOCHS + 1):
    epoch_start = time.time()
    epoch_loss = np.array(0.0) # Initialize as PennyLane numpy array

    # Shuffle data each epoch
    perm = classical_np.random.permutation(len(X_tr))
    X_shuffled = X_tr[perm]
    y_shuffled = y_tr[perm]

    # Mini-batch updates
    for b in range(n_batches):
        start = b * BATCH_SIZE
        end   = start + BATCH_SIZE
        X_b = X_shuffled[start:end]
        y_b = y_shuffled[start:end]

        weights, loss = optimizer.step_and_cost(
            lambda w: cost_fn(w, X_b, y_b),
            weights
        )
        epoch_loss += loss # Removed float()

    avg_loss = epoch_loss / n_batches
    epoch_time = time.time() - epoch_start

    # Training F1 (computed every LOG_EVERY epochs)
    if epoch % LOG_EVERY == 0:
        preds_train = predict(X_tr[:50], weights)  # quick sample
        train_f1 = f1_score(y_tr[:50], preds_train, zero_division=0)
        history['train_f1'].append(train_f1)
        print(f"Epoch {epoch:3d}/{EPOCHS} | Loss: {avg_loss:.4f} | "
              f"Train F1: {train_f1:.3f} | Time: {epoch_time:.1f}s")

    history['loss'].append(avg_loss)
    history['epoch_time'].append(epoch_time)

total_time = time.time() - total_start
print("-" * 55)
print(f"✅ Training complete! Total time: {total_time/60:.1f} min")
print(f"   Final loss: {history['loss'][-1]:.4f}")

# Save weights
classical_np.save('vqc_weights_final.npy', classical_np.array(weights))
print("💾 Saved: vqc_weights_final.npy")
```
